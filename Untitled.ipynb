```python
df['start_year']=df.Span.str.split(pat="-").str[0]
```

```python
df['end_year']=df.Span.str.split(pat="-").str[1]
```

```python
df
```

```python
df['start_year']=df['start_year'].astype(int)
```

```python
df['end_year']=df['end_year'].astype(int)
```

```python
df.info()
```

```python
df['duration']=df['end_year']-df['start_year']
```

```python
df['high_score']=df['high_score'].str.split(pat="*").str[0]
```

```python
df['high_score']=df['high_score'].astype(int)
```

```python
df.info()
```

```python
df['ball_faced']=df['ball_faced'].str.split(pat="+").str[0]
```

```python
df['ball_faced']=df['ball_faced'].replace("-",0)
```

```python
df['strike_rate']=df['strike_rate'].replace("-",0)
```

```python
df['strike_rate']=df['strike_rate'].astype(float)
```
